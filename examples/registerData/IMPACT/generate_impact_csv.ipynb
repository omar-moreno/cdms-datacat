```python
import pandas as pd

# Generates an oldFile,newFile csv for the impact data

# Path of the csv with info on all the files
datalist_path = './data_list.tsv'
# Path of the generated csv
output_path = './copyFiles/fileChecks/IMPACT.csv'
# Path of oldFile
oldfile_path = '/nfs/slac/g/supercdms/tf/northwestern/AnimalData/AR70/processing_coinc_v0701/coincident_TES_PMT_'
# Path of newFile
newfile_path = '/gpfs/slac/staas/fs1/supercdms/data/CDMS/Animal/R70/Processed/Releases/ProdCoinc0701/Merged/'
# Site ID
ID = '27'

# Generates the oldfile name
def oldfile(tes):
    return oldfile_path + str(tes) + '.pkl'
# Generates the newfile name
def newfile(tes):
    # IDyymmdd_hhmmss
    # Convert to string
    tes = str(tes)
    series = ID + tes[2:8] + '_' + tes[8:]
    return newfile_path + series + '.pkl'
    
    

# Read in datalist
datalist_df = pd.read_table(datalist_path)

# Create new dataframe
result_df = pd.concat([pd.DataFrame([[oldfile(tes), newfile(tes)]], columns=['oldFile', 'newFile']) for tes in datalist_df['TES']], ignore_index=True)
result_df.to_csv(output_path)
result_df
```
